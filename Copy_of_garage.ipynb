```python
class Repair:
    """จัดการข้อมูลและการคำนวณงานซ่อมรถ"""

    def __init__(
        self,
        repair_id,
        customer_id,
        vehicle_id,
        repair_type,
        technician_id,
        part_id,
        part_name,
        quantity,
        parts_cost,
        status="Pending"
    ):
        self.repair_id = repair_id
        self.customer_id = customer_id
        self.vehicle_id = vehicle_id
        self.repair_type = repair_type
        self.technician_id = technician_id
        self.part_id = part_id
        self.part_name = part_name
        self.quantity = quantity
        self.parts_cost_value = parts_cost
        self.labor_cost = 0
        self.status = status

    def calculate_labor_cost(self, hours=2, rate_per_hour=300):
        """คำนวณค่าแรง"""

        multiplier = {
            "Engine": 1.5,
            "Brake": 1.2,
            "Suspension": 1.3,
            "Electrical": 1.4,
            "General Check": 1.0,
            "เปลี่ยนน้ำมันเครื่อง": 1.0,
            "เปลี่ยนผ้าเบรก": 1.2,
            "เปลี่ยนแบตเตอรี่": 1.0,
            "ซ่อมเครื่องยนต์": 1.5,
            "ซ่อมระบบไฟ": 1.4
        }

        factor = multiplier.get(self.repair_type, 1.0)

        self.labor_cost = hours * rate_per_hour * factor

        return self.labor_cost

    def calculate_total_price(self):
        """คำนวณราคารวม"""

        return self.parts_cost_value + self.labor_cost

    def approve(self):
        """อนุมัติงานซ่อม"""

        self.status = "Approved"
        return self.status

    def reject(self):
        """ไม่อนุมัติงานซ่อม"""

        self.status = "Rejected"
        return self.status

    def show(self):
        """แสดงข้อมูลงานซ่อม"""

        return {
            "repair_id": self.repair_id,
            "customer_id": self.customer_id,
            "vehicle_id": self.vehicle_id,
            "repair_type": self.repair_type,
            "technician_id": self.technician_id,
            "part_id": self.part_id,
            "part_name": self.part_name,
            "quantity": self.quantity,
            "parts_cost": self.parts_cost_value,
            "labor_cost": self.labor_cost,
            "total_price": self.calculate_total_price(),
            "status": self.status
        }
```

```python
class Inspection:
    """จัดการการตรวจสอบสภาพรถ"""

    def __init__(
        self,
        inspection_id,
        vehicle_id,
        inspector_id,
        result="Pending",
        notes=""
    ):
        self.inspection_id = inspection_id
        self.vehicle_id = vehicle_id
        self.inspector_id = inspector_id
        self.result = result
        self.notes = notes

    def inspect(self, result, notes=""):
        """บันทึกผลการตรวจสอบ"""

        self.result = result
        self.notes = notes

        return self.result

    def pass_inspection(self, notes=""):
        """รถผ่านการตรวจสอบ"""

        self.result = "Passed"
        self.notes = notes

        return self.result

    def fail_inspection(self, notes=""):
        """รถไม่ผ่านการตรวจสอบ"""

        self.result = "Failed"
        self.notes = notes

        return self.result

    def show(self):
        """แสดงข้อมูลการตรวจสอบ"""

        return {
            "inspection_id": self.inspection_id,
            "vehicle_id": self.vehicle_id,
            "inspector_id": self.inspector_id,
            "result": self.result,
            "notes": self.notes
        }
```

```python
import os
import random
import sqlite3
import pandas as pd
from datetime import datetime, timedelta

# Import Module
from approval import ApprovalSystem
from parts import SparePart

# ==========================================
# Core Classes
# ==========================================

class Customer:
    def __init__(self, customer_id, name, phone, address):
        self.customer_id = customer_id
        self.name = name
        self.phone = phone
        self.address = address

class Vehicle:
    def __init__(self, vehicle_id, customer_id, plate_no, brand, model):
        self.vehicle_id = vehicle_id
        self.customer_id = customer_id
        self.plate_no = plate_no
        self.brand = brand
        self.model = model

class ServiceOrder:
    def __init__(self, order_id, vehicle_id, date, status="Pending"):
        self.order_id = order_id
        self.vehicle_id = vehicle_id
        self.date = date
        self.status = status
        self.total_price = 0.0

    def update_status(self, new_status):
        self.status = new_status

    def calculate_total(self, parts_cost, labor_cost):
        self.total_price = parts_cost + labor_cost
        return self.total_price

# ==========================================
# Helper Functions
# ==========================================

def calculate_labor_cost(work_type, hours=2, rate_per_hour=300.0):
    """คำนวณค่าแรงซ่อม (มี Default Argument)"""
    multiplier = {"Engine": 1.5, "Brake": 1.2, "Suspension": 1.3, "Electrical": 1.4, "General Check": 1.0}
    return hours * rate_per_hour * multiplier.get(work_type, 1.0)

def format_service_summary(order_id, status, total_price):
    """จัดรูปแบบข้อความสรุปใบงาน"""
    return f"Order #{order_id} | Status: {status} | Total: {total_price:,.2f} THB"

def validate_qc_pass(inspection_result, threshold=0.85):
    """ตรวจสอบเงื่อนไขผ่าน QC"""
    return inspection_result >= threshold

# ==========================================
# Load Parts Database from CSV
# ==========================================

parts_df = pd.read_csv("auto_parts_inventory.csv")

spare_parts_catalog = []
for idx, row in parts_df.iterrows():
    # รองรับทั้งคอลัมน์ part_id / id หรือราคา unit_price / price
    p_id = row.get("part_id", row.get("id", idx + 101))
    p_name = row.get("part_name", row.get("name", "Unknown Part"))
    p_cat = row.get("category", "General")
    p_price = row.get("unit_price", row.get("price", 500))
    p_supplier = row.get("supplier", "Standard Supplier")

    spare_parts_catalog.append(SparePart(p_id, p_name, p_cat, p_price, p_supplier))

# ==========================================
# Data Generation Loop (จำลอง 304 รายการ)
# ==========================================

random.seed(42) # ล็อก Seed ให้การสุ่มคงเดิมทุกครั้งที่รัน

approval_mgr = ApprovalSystem(owner_name="เจ้าของอู่")

first_names = ["สมชาย", "วิภา", "กิตติ", "นภา", "อนันต์", "ปรียา", "ชาญชัย", "สุภาพร"]
last_names = ["ใจดี", "มีสุข", "วงษ์สุวรรณ", "เจริญสุข", "สมบูรณ์", "รักชาติ"]
brands = ["Toyota", "Honda", "Isuzu", "Nissan", "Mazda"]
models = {"Toyota": "Yaris", "Honda": "Civic", "Isuzu": "D-Max", "Nissan": "Almera", "Mazda": "Mazda 3"}
work_types = ["Engine", "Brake", "Suspension", "Electrical", "General Check"]

records = []
base_date = datetime(2026, 1, 1)

for i in range(1, 305):
    cust_name = f"{random.choice(first_names)} {random.choice(last_names)}"
    phone = f"08{random.randint(10000000, 99999999)}"
    brand = random.choice(brands)
    model = models[brand]
    plate = f"{random.randint(1,9)}กข-{random.randint(1000,9999)}"

    order_date = base_date + timedelta(days=random.randint(0, 50))
    order = ServiceOrder(order_id=i, vehicle_id=i, date=order_date.strftime("%Y-%m-%d"))

    work = random.choice(work_types)
    part_used = random.choice(spare_parts_catalog)
    part_qty = random.randint(1, 4)

    parts_cost = part_used.unit_price * part_qty
    labor_cost = calculate_labor_cost(work_type=work, hours=random.randint(1, 5))
    total_cost = order.calculate_total(parts_cost, labor_cost)

    customer_approved = approval_mgr.customer_approve()

    if customer_approved:
        parts_approved, approval_msg = approval_mgr.request_parts_approval(part_used.name, part_qty, parts_cost)

        if parts_approved:
            part_used.deduct_stock(part_qty)
            qc_score = random.uniform(0.70, 1.0)

            if validate_qc_pass(qc_score):
                order.update_status("Completed & Closed")
                payment_method = random.choice(["Cash", "Transfer", "Credit Card"])
            else:
                order.update_status("Rework Required")
                payment_method = "Pending"
        else:
            order.update_status("Rejected by Owner (Parts)")
            payment_method = "N/A"
    else:
        order.update_status("Cancelled by Customer")
        payment_method = "N/A"

    records.append({
        "order_id": order.order_id,
        "customer_name": cust_name,
        "phone": phone,
        "brand": brand,
        "model": model,
        "plate_no": plate,
        "order_date": order.date,
        "work_type": work,
        "part_id": part_used.part_id,
        "part_name": part_used.name,
        "part_category": part_used.category,
        "part_qty": part_qty,
        "unit_price": part_used.unit_price,
        "parts_cost": parts_cost,
        "labor_cost": labor_cost,
        "total_price": total_cost,
        "status": order.status,
        "payment_method": payment_method
    })

# ==========================================
# Export to CSV & SQLite DB in data_store/
# ==========================================

output_dir = "data_store"
os.makedirs(output_dir, exist_ok=True)

csv_filepath = os.path.join(output_dir, "garage_service_records.csv")
db_filepath = os.path.join(output_dir, "garage_system.db")

df = pd.DataFrame(records)
df.to_csv(csv_filepath, index=False, encoding="utf-8-sig")

# บันทึกลง SQLite DB ทั้งตารางงานซ่อม (service_orders) และตารางฐานข้อมูลอะไหล่ (spare_parts)
conn = sqlite3.connect(db_filepath)
df.to_sql("service_orders", conn, if_exists="replace", index=False)
parts_df.to_sql("spare_parts", conn, if_exists="replace", index=False)
conn.close()

print("จำลองข้อมูลสำเร็จ 304 รายการ!")
print(f"ล็อกไฟล์ CSV ไว้ที่: {csv_filepath}")
print(f"ล็อกไฟล์ Database ไว้ที่: {db_filepath} (รวมตาราง service_orders และ spare_parts)")
```

```python
import os
import random
import sqlite3
import pandas as pd
from datetime import datetime, timedelta


PAYMENT_METHODS = ["Cash", "Transfer", "Credit Card"]

PAYMENT_STATUS_UNPAID = "Unpaid"
PAYMENT_STATUS_PARTIAL = "Partially Paid"
PAYMENT_STATUS_PAID = "Paid"

# ตำแหน่งไฟล์ข้อมูลเดิมของทีม (ตรงกับ main script ล่าสุดที่บันทึกไว้ในโฟลเดอร์ data_store/)
DEFAULT_DATA_DIR = "data_store"
DEFAULT_CSV_NAME = os.path.join(DEFAULT_DATA_DIR, "garage_service_records.csv")
DEFAULT_DB_NAME = os.path.join(DEFAULT_DATA_DIR, "garage_system.db")
DEFAULT_TABLE_NAME = "service_orders"


class Payment:
    """
    การชำระเงิน 1 รายการ สำหรับใบงานซ่อมรถ 1 ใบ (RepairJob)

    Attribute หลัก:
        payment_id     : รหัสการชำระเงิน เช่น "PAY0001"
        order_id       : รหัสใบงานซ่อมที่การชำระเงินนี้อ้างถึง (มาจาก RepairJob.order_id)
        amount_due     : ยอดที่ต้องชำระทั้งหมด (มาจาก job.calculate_total_price())
        payment_method : วิธีชำระเงิน เช่น "Cash", "Transfer", "Credit Card"
        amount_paid    : ยอดที่ลูกค้าชำระมาแล้ว (สะสมได้ถ้าชำระหลายครั้ง)
        payment_date   : วันเวลาที่ชำระเงินล่าสุด
        status         : สถานะการชำระเงิน (Unpaid/Partially Paid/Paid)
    """

    def __init__(self, payment_id, order_id, amount_due, payment_method="Cash"):
        self.payment_id = payment_id
        self.order_id = order_id
        self.amount_due = round(amount_due, 2)
        self.payment_method = payment_method

        self.amount_paid = 0.0
        self.payment_date = None
        self.status = PAYMENT_STATUS_UNPAID

    # ----- เมธอดหลัก -----
    def apply_discount(self, discount_percent=0):
        """ลดยอดที่ต้องชำระตามเปอร์เซ็นต์ส่วนลด ต้องเรียกก่อน process_payment() เท่านั้น"""
        if self.status != PAYMENT_STATUS_UNPAID:
            raise ValueError("ไม่สามารถให้ส่วนลดได้ เนื่องจากมีการชำระเงินไปแล้ว")
        if not (0 <= discount_percent <= 100):
            raise ValueError("เปอร์เซ็นต์ส่วนลดต้องอยู่ระหว่าง 0-100")
        self.amount_due = round(self.amount_due * (1 - discount_percent / 100), 2)

    def process_payment(self, amount_received):
        """รับเงินจากลูกค้า 1 ครั้ง แล้วอัปเดตยอดที่ชำระแล้ว/สถานะ คืนค่าเงินทอน"""
        if amount_received <= 0:
            raise ValueError("จำนวนเงินที่รับต้องมากกว่า 0")

        remaining_before = self.amount_due - self.amount_paid
        change = max(0, amount_received - remaining_before)
        actually_applied = amount_received - change

        self.amount_paid = round(self.amount_paid + actually_applied, 2)
        self.payment_date = datetime.now()

        if self.amount_paid >= self.amount_due:
            self.status = PAYMENT_STATUS_PAID
        else:
            self.status = PAYMENT_STATUS_PARTIAL

        return round(change, 2)

    def is_fully_paid(self):
        """เช็คว่าชำระครบยอดหรือยัง"""
        return self.status == PAYMENT_STATUS_PAID

    def remaining_balance(self):
        """ยอดคงเหลือที่ยังต้องชำระ"""
        return round(max(0, self.amount_due - self.amount_paid), 2)

    def generate_receipt(self):
        """สร้างใบเสร็จแบบ dict สรุปรายการชำระเงิน"""
        return {
            "payment_id": self.payment_id,
            "order_id": self.order_id,
            "payment_method": self.payment_method,
            "amount_due": self.amount_due,
            "amount_paid": self.amount_paid,
            "remaining_balance": self.remaining_balance(),
            "status": self.status,
            "payment_date": self.payment_date.strftime("%Y-%m-%d %H:%M:%S")
                             if self.payment_date else None,
        }

    def to_dict(self):
        """แปลง object เป็น dict เดียว สำหรับนำไปใส่ใน DataFrame/CSV/SQLite ต่อ"""
        return self.generate_receipt()

    def __repr__(self):
        return (f"Payment({self.payment_id}, order={self.order_id}, "
                f"amount_due={self.amount_due}, status={self.status})")


# -----------------------------------------------------------------------
# Helper functions สำหรับจำลองข้อมูลการชำระเงิน
# -----------------------------------------------------------------------
def generate_payment_id(running_number, prefix="PAY"):
    """สร้างรหัสการชำระเงิน เช่น generate_payment_id(1) -> 'PAY0001'"""
    return f"{prefix}{running_number:04d}"


def random_payment_method(methods=None):
    """สุ่มวิธีชำระเงินจาก list ที่กำหนด (default คือ PAYMENT_METHODS)"""
    if methods is None:
        methods = PAYMENT_METHODS
    return random.choice(methods)


def random_payment_date(base_date=None, max_days_ago=30):
    """สุ่มวันที่ชำระเงินย้อนหลังจาก base_date (default = วันนี้) ไม่เกิน max_days_ago วัน"""
    if base_date is None:
        base_date = datetime.now()
    days_ago = random.randint(0, max_days_ago)
    return base_date - timedelta(days=days_ago)


# -----------------------------------------------------------------------
# ฟังก์ชันวิเคราะห์รายได้ / วิธีชำระเงิน ด้วย SQL GROUP BY + SUM + COUNT
# ใช้ฐานข้อมูลเดิม (garage_system.db / service_orders) ที่ main.py สร้างไว้แล้ว
# ไม่ต้องสร้างตารางใหม่ - query ตรงจากคอลัมน์ total_price/payment_method/status
# -----------------------------------------------------------------------
def get_connection(db_name=DEFAULT_DB_NAME):
    """เปิดฐานข้อมูล SQLite เดิมที่ main.py สร้างไว้แล้ว"""
    return sqlite3.connect(db_name)


def _table_exists(conn, table_name):
    """เช็คว่ามีตารางนี้อยู่ในฐานข้อมูลแล้วหรือยัง"""
    row = conn.execute(
        "SELECT name FROM sqlite_master WHERE type='table' AND name=?", (table_name,)
    ).fetchone()
    return row is not None


def load_service_orders(csv_path=DEFAULT_CSV_NAME, db_path=DEFAULT_DB_NAME,
                         table_name=DEFAULT_TABLE_NAME, rebuild_db=False):
    """
    โหลดข้อมูลใบงานซ่อมของทีม โดยยึด 'garage_service_records.csv' เป็นข้อมูลต้นฉบับ
    (ไฟล์ที่ main.py จำลองและบันทึกไว้แล้วใน data_store/) แล้วซิงก์เข้าตาราง
    service_orders ในฐานข้อมูล SQLite (garage_system.db) ก่อนคืน connection กลับไป
    ให้ฟังก์ชันวิเคราะห์ SQL ด้านล่างใช้งานต่อ (ตอบข้อกำหนดที่ต้องมีทั้ง CSV และ SQLite)

    - ถ้ามีไฟล์ฐานข้อมูลอยู่แล้วและมีตาราง service_orders -> ใช้ของเดิมได้เลย (เร็วกว่า)
    - ถ้ายังไม่มี หรือสั่ง rebuild_db=True -> อ่านจาก CSV แล้วสร้างตารางใหม่ทับให้ตรงกับ CSV เสมอ
    """
    db_dir = os.path.dirname(db_path)
    if db_dir:
        os.makedirs(db_dir, exist_ok=True)
    conn = sqlite3.connect(db_path)

    if rebuild_db or not _table_exists(conn, table_name):
        if not os.path.exists(csv_path):
            raise FileNotFoundError(
                f"ไม่พบไฟล์ '{csv_path}' และยังไม่มีตาราง '{table_name}' ในฐานข้อมูล "
                "กรุณารันสคริปต์จำลองข้อมูล (main.py) ก่อน หรือระบุ csv_path ให้ถูกต้อง"
            )
        df = pd.read_csv(csv_path, encoding="utf-8-sig")
        df.to_sql(table_name, conn, if_exists="replace", index=False)
        print(f"ซิงก์ข้อมูลจาก '{csv_path}' เข้าตาราง '{table_name}' ใน '{db_path}' แล้ว ({len(df)} แถว)")

    return conn


def revenue_by_payment_method_sql(conn, table_name=DEFAULT_TABLE_NAME, completed_only=True):
    """
    วิเคราะห์รายได้รวมและจำนวนออเดอร์ แยกตาม 'วิธีชำระเงิน' ด้วย SQL
    ใช้ GROUP BY payment_method, SUM(total_price), COUNT(*)

    completed_only=True (ค่าเริ่มต้น) จะกรองเฉพาะงานที่สถานะ
    'Completed & Closed' เท่านั้น เพราะงานที่ถูก Cancelled/Rejected/
    Rework ยังไม่มีการเก็บเงินจริง (payment_method ยังเป็น 'N/A')
    ถ้าอยากดูยอดใบเสนอราคาทั้งหมดโดยไม่กรอง ให้ส่ง completed_only=False
    """
    where_clause = "WHERE status = 'Completed & Closed'" if completed_only else ""
    query = f"""
        SELECT payment_method,
               COUNT(*)          AS total_orders,
               SUM(total_price)  AS total_revenue,
               AVG(total_price)  AS avg_price
        FROM {table_name}
        {where_clause}
        GROUP BY payment_method
        ORDER BY total_revenue DESC
    """
    return pd.read_sql_query(query, conn)


def revenue_by_work_type_sql(conn, table_name=DEFAULT_TABLE_NAME, completed_only=True):
    """
    วิเคราะห์รายได้รวมและจำนวนงาน แยกตาม 'ประเภทงานซ่อม' (work_type) ด้วย SQL
    ใช้ GROUP BY work_type, SUM(total_price), COUNT(*)

    completed_only=True (ค่าเริ่มต้น) กรองเฉพาะงานที่ปิดจบและเก็บเงินแล้ว
    เพื่อให้ 'รายได้' ที่รายงานเป็นรายได้จริง ไม่ใช่ยอดใบเสนอราคาที่ถูกยกเลิก
    """
    where_clause = "WHERE status = 'Completed & Closed'" if completed_only else ""
    query = f"""
        SELECT work_type,
               COUNT(*)          AS total_orders,
               SUM(total_price)  AS total_revenue,
               AVG(total_price)  AS avg_price
        FROM {table_name}
        {where_clause}
        GROUP BY work_type
        ORDER BY total_revenue DESC
    """
    return pd.read_sql_query(query, conn)


def revenue_by_status_sql(conn, table_name=DEFAULT_TABLE_NAME):
    """
    นับจำนวนออเดอร์และยอดรวม แยกตาม 'สถานะงาน' ด้วย SQL
    ใช้ GROUP BY status, SUM(total_price), COUNT(*)
    (ช่วยดูว่ามีงานถูก Cancelled/Rejected มากน้อยแค่ไหน เทียบกับรายได้จริง)
    """
    query = f"""
        SELECT status,
               COUNT(*)          AS total_orders,
               SUM(total_price)  AS total_revenue
        FROM {table_name}
        GROUP BY status
        ORDER BY total_orders DESC
    """
    return pd.read_sql_query(query, conn)


def revenue_by_payment_method_pandas(df, completed_only=True):
    """
    เวอร์ชัน pandas ของ revenue_by_payment_method_sql() (ใช้ groupby/agg)
    ผลลัพธ์ควรตรงกับเวอร์ชัน SQL ทุกตัวเลข - ต่างกันแค่ไวยากรณ์
    """
    if completed_only:
        df = df[df["status"] == "Completed & Closed"]
    summary = df.groupby("payment_method").agg(
        total_orders=("order_id", "count"),
        total_revenue=("total_price", "sum"),
        avg_price=("total_price", "mean"),
    ).reset_index()
    return summary.sort_values("total_revenue", ascending=False)


# -----------------------------------------------------------------------
# บล็อกทดสอบ: ยึด garage_service_records.csv (ข้อมูลเดิมที่ทีมจำลองไว้แล้ว
# ในโฟลเดอร์ data_store/) เป็นต้นฉบับเสมอ แล้วซิงก์เข้า SQLite ก่อนวิเคราะห์
# ด้วย GROUP BY + SUM + COUNT
# -----------------------------------------------------------------------
if __name__ == "__main__":
    connection = load_service_orders()

    print("\n=== รายได้ แยกตามวิธีชำระเงิน (SQL: GROUP BY + SUM + COUNT) ===")
    print(revenue_by_payment_method_sql(connection))

    print("\n=== รายได้ แยกตามประเภทงานซ่อม (SQL: GROUP BY + SUM + COUNT) ===")
    print(revenue_by_work_type_sql(connection))

    print("\n=== จำนวนออเดอร์/รายได้ แยกตามสถานะงาน (SQL: GROUP BY + SUM + COUNT) ===")
    print(revenue_by_status_sql(connection))

    print("\n=== เทียบผลลัพธ์ด้วย pandas groupby (ควรตรงกับ SQL ข้างบน) ===")
    full_df = pd.read_sql_query(f"SELECT * FROM {DEFAULT_TABLE_NAME}", connection)
    print(revenue_by_payment_method_pandas(full_df))

    connection.close()
```
